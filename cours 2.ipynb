```python
# Boucle for
for i in range(0, 5):   # Je dis à la machine d'afficher le nombre de fois le print dans l'intervalle de 0 à 5
    print("Bonjour")
```

```python
# Boucle while 
i = 0
while i < 3:  # Je dis à la machine d'afficher i tant que i < 10 etant donné que i = 0
    print(i)  # boucle infini
    i = i + 1  # Eviter la boucle la boucle infinie
```
