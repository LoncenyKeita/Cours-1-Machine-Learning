# Produit matricienne
Condition :

Il faut que le nombre de lignes de matrice A soit égal au nbre de colonnes de la matrice B et le rsultat = à la dimension. 

Exemple: A.B = C
    A = m*n et B = n*Z donc remarque ici que dans A le nbre de ligne est n et dans B le nbre de colonne est aussi n donc les 2 valeurs sont égale alors on peut faire le produit matricienne si non il faut d'abord chercher la transposé de B.

Rémarque: Il faut tjrs noter les dimensions de sa matrice ne machine learning
